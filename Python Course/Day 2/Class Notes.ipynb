```python
def is_even(num):
    """ 
    This function returns True if the number is even,
    and it returns False if the number is odd
    """
    if num % 2 == 0:
        return True
    else:
        return False
```

```python
is_even(5)
```

```python
is_even(1222)
```

```python
my_list = [1,2,3,"python",4,5]
print("List sample", my_list)
print("Second element of the list", my_list[1])
my_list[1]=20
my_list.append(True)
print("List sample", my_list)
my_list.remove(my_list[3])
print("List sample", my_list)
```

```python
fruits = ['apple', 'banana', 'cherry']
print(fruits)
fruits.insert(1, "orange")
print(fruits)
```

```python
def insert_at(l, position, e):
    left_part = l[:position]
    left_part.append(e)
    right_part = l[position:]
    return left_part + right_part
```

```python
my_list1=[1,2,3,4,5,6]
new_list = insert_at(my_list1, 3, 999)
print(new_list)
```

# Error Handling

## ZeroDivisionError

```python
x = 3/0 # ZeroDivisionError: division by zero
```

```python
try:
    x=3/0
except ZeroDivisionError as e:
    print(e)
    print("Division by zero can not be performed.")
```

```python
try:
    x=3/0
except Exception as e:
    print(e)
```

## TypeError

```python
try:
    x = 10 + "abc"
except TypeError as e:
    print(f"The following error occured: {e}")
```

## IndexError

```python
my_list = [1,2,3,4]
print(my_list[10]) # IndexError: list index out of range
```

```python
try:
    my_list = [1,2,3,4]
    print(my_list[10])
except Exception as e:
    print(e)
```
