# Hands-on practice: Mastering Python Functions

```python
# Exercise 1
def calculate_area(width, height):
    return width*height

calculate_area(2,5)
```

```python
# Exercise 2
def introduction(name, age=25):
    print(f"This is {name}, and his/her age is {age} years old.")

introduction("Iva",24)
```

```python
# Exercise 3
def first_last(my_list: list):
    first_elem = my_list[0]
    last_elem = my_list[len(my_list)-1]
    print(f"First element is {first_elem}, and last element is {last_elem}")
    return first_elem, last_elem
```

```python
my_list = [1,2,3,4]
first_last(my_list)
```
